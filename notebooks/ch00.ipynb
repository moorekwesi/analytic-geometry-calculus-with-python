## Listing 0.1: Python as a calculator - a Bolt fare in Cape Coast

```python
# Listing 0.1: Python as a calculator - a Bolt fare in Cape Coast
# Illustrative fare structure (NOT official Bolt prices)
base_fare = 6.00        # GH¢ charged as soon as the trip starts
per_km = 2.50           # GH¢ per kilometre
per_min = 0.40          # GH¢ per minute

distance = 7.2          # km, e.g. UCC Science to Kotokuraba Market
time = 18               # minutes

fare = base_fare + per_km * distance + per_min * time
print("Trip fare in GH¢:", fare)
print(f"Rounded to 2 d.p.: GH¢ {fare:.2f}")

# Powers, integer division and remainder
print(2 ** 10)       # 2 to the power 10
print(17 // 5)       # whole-number division
print(17 % 5)        # remainder
```

## Listing 0.2: Lists and for-loops - populations of some regions of Ghana

```python
# Listing 0.2: Lists and for-loops - populations of some regions of Ghana
regions = ["Greater Accra", "Ashanti", "Eastern", "Central", "Northern", "Volta"]
# Populations in thousands, rounded, from the 2021 Population and Housing Census
pop = [5456, 5440, 2926, 2860, 2311, 1659]

for region, p in zip(regions, pop):
    print(f"{region:<14} {p:>6} thousand")

print("Number of regions:", len(regions))
print("Total:", sum(pop), "thousand")
print("Largest:", regions[pop.index(max(pop))])

# A loop that builds a new list: square numbers 1..10
squares = []
for n in range(1, 11):
    squares.append(n ** 2)
print(squares)

# The same thing as a one-line "list comprehension"
print([n ** 2 for n in range(1, 11)])
```

## Listing 0.3: Functions and if-statements - a fare with a minimum charge

```python
# Listing 0.3: Functions and if-statements - a fare with a minimum charge
def bolt_fare(distance_km, minutes, base=6.0, per_km=2.5, per_min=0.4, minimum=15.0):
    """Return an (illustrative) ride-hailing fare in GH¢."""
    fare = base + per_km * distance_km + per_min * minutes
    if fare < minimum:
        fare = minimum
    return round(fare, 2)

print(bolt_fare(1.0, 4))     # a very short trip pays the minimum
print(bolt_fare(7.2, 18))
print(bolt_fare(25, 55))     # Kasoa to Accra Mall, say

# A function of x, written exactly as in mathematics
def f(x):
    return x**2 - 4*x + 3

for x in [0, 1, 2, 3, 4]:
    print("f(", x, ") =", f(x))
```

## Listing 0.4: NumPy arrays - computing many values at once

```python
# Listing 0.4: NumPy arrays - computing many values at once
import numpy as np

x = np.array([0, 1, 2, 3, 4])
print("x       =", x)
print("x**2    =", x**2)            # squares every entry
print("2*x + 1 =", 2*x + 1)

# 5 equally spaced points from 0 to 1 (inclusive)
print(np.linspace(0, 1, 5))

# Built-in mathematical functions act on whole arrays
t = np.linspace(0, np.pi, 5)
print("sin(t) =", np.round(np.sin(t), 4))
print("pi =", np.pi, " e =", np.e, " sqrt(2) =", np.sqrt(2))

# Summary statistics of daily Bolt trips for a driver (illustrative)
trips = np.array([14, 17, 12, 20, 22, 25, 9])   # Mon..Sun
print("mean =", trips.mean(), " max =", trips.max(), " total =", trips.sum())
```

## Listing 0.5: Plotting with Matplotlib - comparing two fare plans

```python
# Listing 0.5: Plotting with Matplotlib - comparing two fare plans
import numpy as np
import matplotlib.pyplot as plt

d = np.linspace(0, 30, 200)          # distance in km
plan_A = 6.0 + 2.5 * d               # higher per-km rate
plan_B = 12.0 + 2.0 * d              # higher base fare, cheaper per km

plt.figure(figsize=(6, 3.6))
plt.plot(d, plan_A, label="Plan A: 6 + 2.5d")
plt.plot(d, plan_B, "--", label="Plan B: 12 + 2.0d")
plt.plot(12, 36, "ko")               # the plans cost the same at d = 12 km
plt.annotate("equal cost at 12 km", (12, 36), (15, 25), arrowprops=dict(arrowstyle="->"))
plt.xlabel("distance d (km)")
plt.ylabel("fare (GH¢)")
plt.title("Two illustrative fare plans")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Listing 0.6: SymPy - exact (symbolic) mathematics

```python
# Listing 0.6: SymPy - exact (symbolic) mathematics
import sympy as sp

x = sp.symbols("x")
f = x**2 - 4*x + 3

print("f(x)       =", f)
print("factorised =", sp.factor(f))
print("roots      =", sp.solve(f, x))
print("(x+1)^3    =", sp.expand((x + 1)**3))
print("f(5)       =", f.subs(x, 5))

# Exact arithmetic: no rounding errors
print(sp.Rational(1, 3) + sp.Rational(1, 6))
print(sp.sqrt(8))

# A preview of the calculus to come
print("limit of sin(x)/x as x->0:", sp.limit(sp.sin(x)/x, x, 0))
print("derivative of f:", sp.diff(f, x))
print("integral of f:", sp.integrate(f, x))
```
