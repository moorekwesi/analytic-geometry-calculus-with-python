## Listing 1.1: The distance formula in Python

```python
# Listing 1.1: The distance formula in Python
import math

def distance(A, B):
    """Distance between points A=(x1, y1) and B=(x2, y2)."""
    (x1, y1), (x2, y2) = A, B
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Checking the worked example: which of P or Q is closer to A?
A, P, Q = (3, 5), (-1, -2), (8, 9)
print(f"d(P,A) = {distance(P, A):.4f}  (sqrt(65) = {math.sqrt(65):.4f})")
print(f"d(Q,A) = {distance(Q, A):.4f}  (sqrt(41) = {math.sqrt(41):.4f})")

# Approximate positions (km) on a simple grid of Cape Coast:
# origin at Cape Coast Castle, x-axis pointing east, y-axis pointing north.
places = {
    "Cape Coast Castle": (0.0, 0.0),
    "Kotokuraba Market": (-0.4, 0.5),
    "UCC Science":       (-4.4, 1.5),
    "Abura":             (-1.6, 2.4),
}
start = "UCC Science"
for name, pt in places.items():
    if name != start:
        print(f"{start} -> {name:<18}: {distance(places[start], pt):5.2f} km")
```

## Listing 1.2: Flat-map distance versus true (great-circle) distance in Ghana

```python
# Listing 1.2: Flat-map distance versus true (great-circle) distance in Ghana
import math

R = 6371.0   # mean radius of the Earth in km
towns = {    # (latitude, longitude) in degrees; west is negative
    "Accra":      (5.6037, -0.1870),
    "Kumasi":     (6.6885, -1.6244),
    "Cape Coast": (5.1053, -1.2466),
    "Tamale":     (9.4008, -0.8393),
}

def flat_distance(P, Q):
    """Treat (lat, lon) as x-y coordinates on a flat map.
    East-west distances are shrunk by cos(latitude) because meridians converge."""
    lat0 = math.radians((P[0] + Q[0]) / 2)
    dx = math.radians(Q[1] - P[1]) * R * math.cos(lat0)
    dy = math.radians(Q[0] - P[0]) * R
    return math.sqrt(dx**2 + dy**2)        # the distance formula!

def great_circle(P, Q):
    """Haversine formula: distance along the surface of a sphere."""
    p1, p2 = math.radians(P[0]), math.radians(Q[0])
    dphi = p2 - p1
    dlam = math.radians(Q[1] - P[1])
    a = math.sin(dphi/2)**2 + math.cos(p1)*math.cos(p2)*math.sin(dlam/2)**2
    return 2 * R * math.asin(math.sqrt(a))

for B in ["Kumasi", "Cape Coast", "Tamale"]:
    f = flat_distance(towns["Accra"], towns[B])
    g = great_circle(towns["Accra"], towns[B])
    print(f"Accra -> {B:<10}  flat map: {f:6.1f} km   sphere: {g:6.1f} km")
```

## Listing 1.3: Midpoints and the parallelogram PQRS

```python
# Listing 1.3: Midpoints and the parallelogram PQRS
import matplotlib.pyplot as plt

def midpoint(A, B):
    return ((A[0] + B[0]) / 2, (A[1] + B[1]) / 2)

P, Q, R, S = (1, 2), (4, 4), (5, 9), (2, 7)
M1, M2 = midpoint(P, R), midpoint(Q, S)
print("Midpoint of diagonal PR:", M1)
print("Midpoint of diagonal QS:", M2)
print("Diagonals bisect each other:", M1 == M2)

xs = [P[0], Q[0], R[0], S[0], P[0]]
ys = [P[1], Q[1], R[1], S[1], P[1]]
plt.figure(figsize=(4, 4.5))
plt.plot(xs, ys, "b-o")
plt.plot([P[0], R[0]], [P[1], R[1]], "g--")
plt.plot([Q[0], S[0]], [Q[1], S[1]], "g--")
plt.plot(*M1, "rs")
for name, pt in zip("PQRS", [P, Q, R, S]):
    plt.annotate(name, pt, textcoords="offset points", xytext=(6, -10))
plt.annotate("M(3, 5.5)", M1, textcoords="offset points", xytext=(8, 0))
plt.gca().set_aspect("equal")
plt.grid(alpha=0.3)
plt.show()
```

## Listing 1.4: Gradients - parallel and perpendicular lines, and road slopes

```python
# Listing 1.4: Gradients - parallel and perpendicular lines, and road slopes
from fractions import Fraction
import math

def gradient(A, B):
    (x1, y1), (x2, y2) = A, B
    return Fraction(y2 - y1, x2 - x1)   # exact fraction, no rounding

A, B, C, D = (1, 2), (4, 8), (0, 5), (2, 4)
mAB, mCD = gradient(A, B), gradient(C, D)
print("m_AB =", mAB, "  m_CD =", mCD)
print("Parallel?      ", mAB == mCD)
print("Perpendicular? ", mAB * mCD == -1)

# A road rises 45 m over a horizontal distance of 600 m
rise, run = 45, 600
m = rise / run
print(f"gradient = {m:.3f}, i.e. a {100*m:.1f}% grade")
print(f"angle of inclination = {math.degrees(math.atan(m)):.2f} degrees")
```

## Listing 1.5: Finding a fare formula (a straight line) from two trips

```python
# Listing 1.5: Finding a fare formula (a straight line) from two trips
import sympy as sp

x, y = sp.symbols("x y")

def line_through(P, Q):
    """Return the line through P and Q in the form y = m x + c."""
    m = sp.Rational(Q[1] - P[1], Q[0] - P[0])
    return sp.Eq(y, sp.expand(m * (x - P[0]) + P[1]))

# Checking the worked example: line through (1, 3) with gradient 5
print(sp.Eq(y, sp.expand(5 * (x - 1) + 3)))

# Two trips with the same driver (distance km, fare GH¢), illustrative
trip1, trip2 = (4, 18), (10, 33)
L = line_through(trip1, trip2)
print("Fare line:", L)
m = L.rhs.coeff(x)
c = L.rhs.subs(x, 0)
print("per-km rate =", m, "  base fare =", c)
print("Predicted fare for a 15 km trip:", L.rhs.subs(x, 15))

# The same line in general form ax + by + c = 0 with integer coefficients
num, den = sp.fraction(sp.together(L.rhs - y))
print("General form:", sp.Eq(sp.expand(num), 0))
```

## Listing 1.6: Angle between two lines

```python
# Listing 1.6: Angle between two lines
import math

def acute_angle(m1, m2):
    """Acute angle (degrees) between lines with gradients m1 and m2."""
    if 1 + m1 * m2 == 0:
        return 90.0
    return math.degrees(math.atan(abs((m1 - m2) / (1 + m1 * m2))))

print(f"y=2x+1 and y=3x+2: {acute_angle(3, 2):.4f} degrees")

# Two roads meet at a junction: one along y = 0.5x, the other along y = -3x + 14.
print(f"Junction angle: {acute_angle(0.5, -3):.1f} degrees")

# Check with angles of inclination: theta = atan(m)
t1, t2 = math.atan(3), math.atan(2)
print(f"Check: {math.degrees(t1 - t2):.4f} degrees")
```

## Listing 1.7: The parallelogram ABCD solved with SymPy

```python
# Listing 1.7: The parallelogram ABCD solved with SymPy
import sympy as sp

x, y, t = sp.symbols("x y t")
A = sp.Point(1, 2)
AB = sp.Line(sp.Point(-3, 0), sp.Point(1, 2))      # 2y = x + 3 passes through these
BD = sp.Line(sp.Point(17, 0), sp.Point(2, 5))      # 3y + x = 17 passes through these

B = AB.intersection(BD)[0]
print("B =", B)

# D is the point of BD (other than B) with |AD| = |BD|.
D_general = sp.Point(17 - 3*t, t)                 # every point of BD: x = 17 - 3y
sols = sp.solve(sp.Eq(A.distance(D_general)**2, B.distance(D_general)**2), t)
D = D_general.subs(t, sols[0])
print("D =", D)

M = B.midpoint(D)                                 # diagonals bisect each other
C = sp.Point(2*M.x - A.x, 2*M.y - A.y)
print("M =", M, "  C =", C)

mid_AB = A.midpoint(B)
perp = AB.perpendicular_line(mid_AB)
print("Perpendicular bisector of AB:", sp.expand(perp.equation(x, y)), "= 0")
```

## Listing 1.8: Dividing a line segment in a given ratio

```python
# Listing 1.8: Dividing a line segment in a given ratio
from fractions import Fraction as F

def divide(A, B, n, m, external=False):
    """Point R with AR:RB = n:m (internal) or n:-m (external)."""
    (x1, y1), (x2, y2) = A, B
    if external:
        m = -m
    return (F(m*x1 + n*x2, m + n), F(m*y1 + n*y2, m + n))

A, B = (1, 2), (3, 1)
R = divide(A, B, 1, 2)
print("Internal 1:2 ->", tuple(str(v) for v in R))
S = divide(A, B, 1, 2, external=True)
print("External 1:2 ->", tuple(str(v) for v in S))

# A water pipeline runs from a treatment plant W(0, 0) to a town T(12, 9) (km).
# A booster pump must be placed at P with WP : PT = 2 : 3.
P = divide((0, 0), (12, 9), 2, 3)
print("Pump position:", tuple(float(v) for v in P))
```

## Listing 1.9: Shortest feeder road from a village to a straight highway

```python
# Listing 1.9: Shortest feeder road from a village to a straight highway
import math
import matplotlib.pyplot as plt
import numpy as np

def point_line_distance(P, a, b, c):
    """Distance from P=(x1, y1) to the line ax + by + c = 0."""
    x1, y1 = P
    return abs(a*x1 + b*y1 + c) / math.sqrt(a**2 + b**2)

def foot_of_perpendicular(P, a, b, c):
    x1, y1 = P
    k = (a*x1 + b*y1 + c) / (a**2 + b**2)
    return (x1 - a*k, y1 - b*k)

# Checking the worked example: P(3, 2) and 12x + 5y - 11 = 0
print("Distance =", point_line_distance((3, 2), 12, 5, -11))

# Highway: 3x + 4y - 12 = 0 (km grid).  Village V(6, 8).
V = (6, 8)
d = point_line_distance(V, 3, 4, -12)
F = foot_of_perpendicular(V, 3, 4, -12)
print(f"Feeder road length = {d:.2f} km, meeting the highway at ({F[0]:.2f}, {F[1]:.2f})")

x = np.linspace(-2, 8, 50)
plt.figure(figsize=(5, 4.5))
plt.plot(x, (12 - 3*x)/4, "k", lw=3, label="highway 3x + 4y = 12")
plt.plot([V[0], F[0]], [V[1], F[1]], "g--o", label=f"feeder road, {d:.1f} km")
plt.annotate("village V", V, textcoords="offset points", xytext=(5, 5))
plt.gca().set_aspect("equal")
plt.grid(alpha=0.3)
plt.legend(loc="lower left")
plt.show()
```

## Listing 1.10: Bisectors of the angles between two lines (with SymPy)

```python
# Listing 1.10: Bisectors of the angles between two lines (with SymPy)
import sympy as sp

x, y = sp.symbols("x y")
L1 = x + y + 2
L2 = x + 7*y + 26

# A point on a bisector is equally far from both lines:
#     L1 / sqrt(1**2 + 1**2) = +/- L2 / sqrt(1**2 + 7**2)
for sign in (+1, -1):
    eq = sp.expand(L1 / sp.sqrt(2) - sign * L2 / sp.sqrt(50))
    eq = sp.expand(eq * 5 * sp.sqrt(2))      # clear the surds
    print("Bisector:", eq, "= 0   i.e.", sp.factor_terms(eq), "= 0")

# Borehole problem: two straight roads y = 0 and 4x - 3y = 0 meet at a junction O.
# The borehole must be equidistant from both roads, inside the angle between them
# (x > 0, y > 0), and 5 km from the junction.
bisector = sp.Eq(y, (4*x - 3*y) / 5)         # distances y and (4x-3y)/5 are equal
sol = sp.solve([bisector, sp.Eq(x**2 + y**2, 25)], [x, y], dict=True)
for s in sol:
    if s[x] > 0:
        print("Borehole at", (s[x], s[y]),
              "~", (round(float(s[x]), 3), round(float(s[y]), 3)))
```
