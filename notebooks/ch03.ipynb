## Listing 3.1: Guessing a limit from a table of values

```python
# Listing 3.1: Guessing a limit from a table of values
import math

def f(x):
    return (x**2 - 16) / (x - 4)      # undefined at x = 4

def g(x):
    return math.sin(x) / x            # undefined at x = 0 (x in radians)

print(f"{'h':>10} {'f(4-h)':>12} {'f(4+h)':>12} {'g(-h)':>12} {'g(h)':>12}")
for k in range(1, 7):
    h = 10**(-k)
    print(f"{h:>10.0e} {f(4-h):>12.6f} {f(4+h):>12.6f} {g(-h):>12.8f} {g(h):>12.8f}")
```

## Listing 3.2: From average speed to instantaneous speed

```python
# Listing 3.2: From average speed to instantaneous speed
# A Bolt car pulls away from the traffic lights at Kotokuraba.
# Distance travelled (metres) after t seconds:  s(t) = 1.5 t^2   (0 <= t <= 10)

def s(t):
    return 1.5 * t**2

t0 = 4.0
print(f"{'interval':>22} {'average speed (m/s)':>20}")
for h in [1, 0.5, 0.1, 0.01, 0.001, 0.0001]:
    avg = (s(t0 + h) - s(t0)) / h
    print(f"[{t0}, {t0 + h:<9}] {avg:>20.6f}")
print("The averages approach 12 m/s =", 12 * 3.6, "km/h: the speedometer reading at t = 4 s.")
```

## Listing 3.3: Checking the worked examples with sp.limit

```python
# Listing 3.3: Checking the worked examples with sp.limit
import sympy as sp

x, t = sp.symbols("x t")
examples = [
    ("2x^2 + 1, x->1",                2*x**2 + 1,                      x, 1),
    ("(x^2-2x+1)/(3x-2), x->3",       (x**2 - 2*x + 1)/(3*x - 2),      x, 3),
    ("(x^2-16)/(x-4), x->4",          (x**2 - 16)/(x - 4),             x, 4),
    ("(sqrt(x+9)-3)/x, x->0",         (sp.sqrt(x + 9) - 3)/x,          x, 0),
    ("tan(x)/x, x->0",                sp.tan(x)/x,                     x, 0),
    ("sin(4x)/x, x->0",               sp.sin(4*x)/x,                   x, 0),
    ("(cos3x - cos x)/x^2, x->0",     (sp.cos(3*x) - sp.cos(x))/x**2,  x, 0),
    ("(1 - cos x)/x^2, x->0",         (1 - sp.cos(x))/x**2,            x, 0),
]
for name, expr, var, a in examples:
    print(f"{name:<28} = {sp.limit(expr, var, a)}")

# One-sided limits
expr = sp.Abs(x - 8)/(x - 8)
print("|x-8|/(x-8): left =", sp.limit(expr, x, 8, dir="-"), ", right =", sp.limit(expr, x, 8, dir="+"))
```

## Listing 3.4: sin(x)/x near 0 - and why radians matter

```python
# Listing 3.4: sin(x)/x near 0 - and why radians matter
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)
x = x[x != 0]                              # avoid dividing by zero
plt.figure(figsize=(6, 3.2))
plt.plot(x, np.sin(x)/x, label="sin(x)/x")
plt.plot(0, 1, "o", mfc="white", mec="C0")   # the 'hole' at x = 0
plt.axhline(0, color="k", lw=0.5)
plt.grid(alpha=0.3); plt.legend()
plt.show()

# If x is measured in DEGREES the limit is not 1:
for d in [10, 1, 0.1, 0.01]:
    print(f"x = {d:>5} degrees:  sin(x)/x = {np.sin(np.radians(d))/d:.8f}")
print("pi/180 =", np.pi/180)
```

## Listing 3.5: One-sided limits of a tariff with a jump (illustrative prices)

```python
# Listing 3.5: One-sided limits of a tariff with a jump (illustrative prices)
# Monthly electricity bill B(u) in GH¢ for u units (kWh):
#   up to 30 units: 0.65 per unit (a "lifeline" band, no service charge)
#   above 30 units: 1.20 per unit plus a fixed service charge of GH¢ 10
import numpy as np
import matplotlib.pyplot as plt

def bill(u):
    return np.where(u <= 30, 0.65*u, 10 + 1.20*u)

for u in [29.9, 29.99, 29.999, 30, 30.001, 30.01, 30.1]:
    print(f"B({u}) = {float(bill(u)):.4f}")
print("Left-hand limit at 30  :", 0.65*30)
print("Right-hand limit at 30 :", 10 + 1.20*30)

u1 = np.linspace(0, 30, 50); u2 = np.linspace(30.0001, 60, 50)
plt.figure(figsize=(5.5, 3.4))
plt.plot(u1, bill(u1), "C0"); plt.plot(u2, bill(u2), "C0")
plt.plot(30, 19.5, "o", color="C0"); plt.plot(30, 46, "o", mfc="white", mec="C0")
plt.xlabel("units used (kWh)"); plt.ylabel("bill (GH¢)"); plt.grid(alpha=0.3)
plt.show()
```

## Listing 3.7: The squeeze (sandwich) theorem: -x^2 <= x^2 sin(1/x) <= x^2

```python
# Listing 3.7: The squeeze (sandwich) theorem: -x^2 <= x^2 sin(1/x) <= x^2
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-0.4, 0.4, 4001)
x = x[x != 0]
plt.figure(figsize=(5.5, 3.4))
plt.plot(x, x**2, "k--", lw=1, label="$x^2$")
plt.plot(x, -x**2, "k--", lw=1, label="$-x^2$")
plt.plot(x, x**2*np.sin(1/x), "C3", lw=0.8, label="$x^2\\sin(1/x)$")
plt.legend(); plt.grid(alpha=0.3)
plt.show()

for h in [0.1, 0.01, 0.001]:
    print(f"x = {h}:  x^2 sin(1/x) = {h**2*np.sin(1/h):+.2e}  (between ±{h**2:.0e})")
```

## Listing 3.8: Three ways a function can fail to be continuous

```python
# Listing 3.8: Three ways a function can fail to be continuous
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 3, figsize=(9, 2.8))

# (a) a removable discontinuity (hole): (x^2 - 4)/(x - 2)
x = np.linspace(0, 4, 200); x = x[np.abs(x - 2) > 1e-9]
ax[0].plot(x, (x**2 - 4)/(x - 2)); ax[0].plot(2, 4, "o", mfc="white", mec="C0")
ax[0].set_title("(a) hole at x = 2")

# (b) a jump: f(x) = x for x < 1, x + 2 for x >= 1
x1 = np.linspace(-1, 1, 50, endpoint=False); x2 = np.linspace(1, 3, 50)
ax[1].plot(x1, x1, "C0"); ax[1].plot(x2, x2 + 2, "C0")
ax[1].plot(1, 1, "o", mfc="white", mec="C0"); ax[1].plot(1, 3, "o", color="C0")
ax[1].set_title("(b) jump at x = 1")

# (c) an infinite discontinuity (vertical asymptote): 1/(x - 1)^2
x = np.linspace(-1, 3, 400); x = x[np.abs(x - 1) > 0.08]
y = 1/(x - 1)**2
ax[2].plot(x[x < 1], y[x < 1], "C0"); ax[2].plot(x[x > 1], y[x > 1], "C0")
ax[2].axvline(1, ls="--", color="gray"); ax[2].set_ylim(0, 40)
ax[2].set_title("(c) asymptote at x = 1")

for a in ax:
    a.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Listing 3.9: The Intermediate Value Theorem at work - the bisection method

```python
# Listing 3.9: The Intermediate Value Theorem at work - the bisection method
def bisection(f, a, b, tol=1e-6):
    """Find a root of a continuous f on [a, b], where f(a) and f(b) differ in sign."""
    if f(a) * f(b) > 0:
        raise ValueError("f(a) and f(b) must have opposite signs")
    steps = 0
    while b - a > tol:
        m = (a + b) / 2
        if f(a) * f(m) <= 0:      # sign change in the left half
            b = m
        else:                      # sign change in the right half
            a = m
        steps += 1
    return (a + b) / 2, steps

f = lambda x: x**3 - x - 2
root, n = bisection(f, 1, 2)
print(f"root of x^3 - x - 2 = 0: {root:.6f}  ({n} halvings)")

# A kenkey seller's daily profit (GH¢) when she makes x hundred balls (illustrative):
#   P(x) = -4x^3 + 30x^2 - 20x - 15.  When does she first break even (P = 0)?
P = lambda x: -4*x**3 + 30*x**2 - 20*x - 15
print("P(0) =", P(0), ", P(2) =", P(2))
root, n = bisection(P, 0, 2)
print(f"break-even at about {100*root:.0f} balls")
```

## Listing 3.10: Epsilon and delta - how close is close enough?

```python
# Listing 3.10: Epsilon and delta - how close is close enough?
# lim_{x->2} x^2 = 4. Given a tolerance eps, find delta so that
# |x - 2| < delta guarantees |x^2 - 4| < eps.
import numpy as np

def works(delta, eps, n=100001):
    x = np.linspace(2 - delta, 2 + delta, n)[1:-1]
    return np.all(np.abs(x**2 - 4) < eps)

for eps in [1, 0.1, 0.01, 0.001]:
    delta = 1.0
    while not works(delta, eps):
        delta /= 2                       # halve delta until it works
    rule = min(1, eps/5)                 # the textbook choice, proved in the text
    print(f"eps = {eps:<6} found delta = {delta:<12.6g} textbook delta = min(1, eps/5) = {rule:g}")
```
