## Listing 6.1: Indefinite integrals with sp.integrate (SymPy omits the + C)

```python
# Listing 6.1: Indefinite integrals with sp.integrate (SymPy omits the + C)
import sympy as sp

x = sp.symbols("x", positive=True)
examples = [
    3*x**2, sp.cos(x), x**8, 1/x**3, sp.cbrt(x**2), sp.tan(x)/sp.sec(x),
    5*x**3 + 2*sp.cos(x), 8*x**3 - 6*sp.sqrt(x) + 1/x**3, (x**2 - 1)**2/x**2,
    1/(sp.cos(x)*sp.cot(x)),
]
for f in examples:
    F = sp.integrate(f, x)
    ok = sp.simplify(sp.diff(F, x) - f) == 0      # always check: F' = f
    print(f"∫ {str(f):<28} dx = {str(sp.simplify(F)):<34} check: {ok}")
```

## Listing 6.2: Integration by substitution, checked by differentiating

```python
# Listing 6.2: Integration by substitution, checked by differentiating
import sympy as sp

x = sp.symbols("x", positive=True)
hand_answers = {                                   # integrand: our answer by hand
    sp.sqrt(5*x + 7):                 sp.Rational(2, 15)*(5*x + 7)**sp.Rational(3, 2),
    (x**2 - 1)/(x**3 - 3*x + 1)**6:   -1/(15*(x**3 - 3*x + 1)**5),
    sp.cos(5*x)**3*sp.sin(5*x):       -sp.cos(5*x)**4/20,
    sp.exp(sp.sqrt(x))/sp.sqrt(x):    2*sp.exp(sp.sqrt(x)),
    sp.log(x)/x:                      sp.log(x)**2/2,
    sp.cos(4*x):                      sp.sin(4*x)/4,
    (2*x**3 + 1)**7*x**2:             (2*x**3 + 1)**8/48,
}
for f, F in hand_answers.items():
    ok = sp.simplify(sp.diff(F, x) - f) == 0
    print(f"∫ {str(f):<32} dx = {str(F):<30} + C   correct: {ok}")
```

## Listing 6.3: Riemann sums for the area under y = x^2 on [0, 1]

```python
# Listing 6.3: Riemann sums for the area under y = x^2 on [0, 1]
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: x**2
a, b = 0, 1

def riemann(f, a, b, n, rule="left"):
    dx = (b - a) / n
    left = a + dx*np.arange(n)                 # left endpoints x_0 ... x_{n-1}
    pts = {"left": left, "right": left + dx, "mid": left + dx/2}[rule]
    return np.sum(f(pts)) * dx

print(f"{'n':>6} {'left':>10} {'right':>10} {'midpoint':>10}")
for n in [4, 10, 100, 1000, 10000]:
    print(f"{n:>6} {riemann(f, a, b, n):>10.6f} {riemann(f, a, b, n, 'right'):>10.6f}"
          f" {riemann(f, a, b, n, 'mid'):>10.6f}")
print("exact area = 1/3 =", 1/3)

# Picture of the right-endpoint rectangles for n = 8
n = 8; dx = (b - a)/n; xr = a + dx*np.arange(1, n + 1)
X = np.linspace(a, b, 200)
plt.figure(figsize=(5, 3.4))
plt.bar(xr - dx, f(xr), width=dx, align="edge", alpha=0.35, edgecolor="k", label="right sums, n = 8")
plt.plot(X, f(X), "C3", lw=2, label="$y = x^2$")
plt.legend(); plt.grid(alpha=0.3); plt.show()
```

## Listing 6.4: Definite integrals and the Fundamental Theorem of Calculus

```python
# Listing 6.4: Definite integrals and the Fundamental Theorem of Calculus
import sympy as sp

x, t = sp.symbols("x t")
print("∫_1^2 x dx                =", sp.integrate(x, (x, 1, 2)))
print("∫_2^10 3/sqrt(5x-1) dx    =", sp.integrate(3/sp.sqrt(5*x - 1), (x, 2, 10)))
print("∫_0^pi sin x dx           =", sp.integrate(sp.sin(x), (x, 0, sp.pi)))
print("∫_0^2pi sin x dx          =", sp.integrate(sp.sin(x), (x, 0, 2*sp.pi)), " (signed area!)")

# FTC part 1: d/dx of ∫_0^x f(t) dt is f(x)
G = sp.integrate(t**3 + 1, (t, 0, x))
print("G(x) = ∫_0^x (t^3 + 1) dt =", G, ";  G'(x) =", sp.diff(G, x))

# Archimedes: the parabolic segment cut from y = x^2 by y = 1 has area 4/3 of
# the triangle with the same base and vertex.
segment = sp.integrate(1 - x**2, (x, -1, 1))
triangle = sp.Rational(1, 2) * 2 * 1
print("segment =", segment, ", triangle =", triangle, ", ratio =", segment/triangle)
```

## Listing 6.5: Integration by parts, checked with SymPy

```python
# Listing 6.5: Integration by parts, checked with SymPy
import sympy as sp

x = sp.symbols("x", positive=True)

# Our answers by hand, using  ∫ u dv = u v - ∫ v du
hand_answers = {
    x*sp.exp(x):        (x - 1)*sp.exp(x),
    x*sp.cos(x):        x*sp.sin(x) + sp.cos(x),
    sp.log(x):          x*sp.log(x) - x,
    x**2*sp.exp(x):     (x**2 - 2*x + 2)*sp.exp(x),
    sp.exp(x)*sp.sin(x): sp.exp(x)*(sp.sin(x) - sp.cos(x))/2,
}
for f, F in hand_answers.items():
    ok = sp.simplify(sp.diff(F, x) - f) == 0
    print(f"∫ {str(f):<16} dx = {str(F):<32} + C   correct: {ok}")

# A definite integral by parts
print("∫_0^pi x sin x dx =", sp.integrate(x*sp.sin(x), (x, 0, sp.pi)))
```

## Listing 6.6: Trapezium rule versus Simpson's rule on ∫_0^pi sin x dx = 2

```python
# Listing 6.6: Trapezium rule versus Simpson's rule on ∫_0^pi sin x dx = 2
import numpy as np

def trapezium(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    h = (b - a) / n
    y = f(x)
    return h/2 * (y[0] + 2*y[1:-1].sum() + y[-1])

def simpson(f, a, b, n):                 # n must be even
    x = np.linspace(a, b, n + 1)
    h = (b - a) / n
    y = f(x)
    return h/3 * (y[0] + 4*y[1:-1:2].sum() + 2*y[2:-1:2].sum() + y[-1])

print(f"{'n':>4} {'trapezium':>12} {'error':>10} {'Simpson':>12} {'error':>10}")
for n in [2, 4, 8, 16, 32]:
    T, S = trapezium(np.sin, 0, np.pi, n), simpson(np.sin, 0, np.pi, n)
    print(f"{n:>4} {T:>12.8f} {abs(T - 2):>10.2e} {S:>12.8f} {abs(S - 2):>10.2e}")
```

## Listing 6.7: Integrals with no formula - scipy.integrate.quad

```python
# Listing 6.7: Integrals with no formula - scipy.integrate.quad
import numpy as np
from scipy.integrate import quad

# exp(-x^2) has no elementary antiderivative, but its integral is easy to compute
val, err = quad(lambda x: np.exp(-x**2), 0, 1)
print(f"∫_0^1 exp(-x^2) dx = {val:.10f}  (error estimate {err:.1e})")

# quad also handles infinite limits
val, _ = quad(lambda x: np.exp(-x**2), -np.inf, np.inf)
print(f"∫ over all x       = {val:.10f};  sqrt(pi) = {np.sqrt(np.pi):.10f}")

# and integrands such as sin(x)/x
val, _ = quad(lambda x: np.sinc(x/np.pi), 0, 10)       # np.sinc(t) = sin(pi t)/(pi t)
print(f"∫_0^10 sin(x)/x dx = {val:.10f}")
```
