## Listing 7.1: Area between two curves: y = 2x and y = x^2

```python
# Listing 7.1: Area between two curves: y = 2x and y = x^2
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

x = sp.symbols("x")
top, bottom = 2*x, x**2
a, b = sp.solve(sp.Eq(top, bottom), x)          # intersection points
area = sp.integrate(top - bottom, (x, a, b))
print("curves meet at x =", a, "and x =", b, ";  area =", area)

X = np.linspace(-0.3, 2.3, 200)
plt.figure(figsize=(4.5, 3.6))
plt.plot(X, 2*X, label="$y = 2x$"); plt.plot(X, X**2, label="$y = x^2$")
Xf = np.linspace(0, 2, 100)
plt.fill_between(Xf, Xf**2, 2*Xf, alpha=0.3, label=f"area = {area}")
plt.legend(); plt.grid(alpha=0.3); plt.show()
```

## Listing 7.2: Distance travelled from speedometer readings (trapezium rule)

```python
# Listing 7.2: Distance travelled from speedometer readings (trapezium rule)
# A trotro's speed (km/h) read every 5 minutes on the Kasoa - Accra road (illustrative)
import numpy as np

t_min = np.array([0, 5, 10, 15, 20, 25, 30, 35, 40, 45])
speed = np.array([0, 18, 35, 12, 8, 30, 52, 47, 25, 0])     # km/h
t_hr = t_min / 60

distance = np.trapz(speed, t_hr)                           # ∫ v dt
print(f"distance travelled ≈ {distance:.2f} km in {t_min[-1]} minutes")
print(f"average speed     ≈ {distance/t_hr[-1]:.1f} km/h")

# Same thing 'by hand': each strip is a trapezium with area h*(v_i + v_{i+1})/2
h = 5/60
by_hand = sum(h*(speed[i] + speed[i + 1])/2 for i in range(len(speed) - 1))
print(f"by hand           ≈ {by_hand:.2f} km")
```

## Listing 7.3: Total rainfall from rainfall intensity (an accumulation problem)

```python
# Listing 7.3: Total rainfall from rainfall intensity (an accumulation problem)
# Intensity r(t) in mm/hour during an afternoon storm in Kumasi, every 10 min (illustrative)
import numpy as np

t = np.arange(0, 121, 10) / 60                      # hours: 0, 1/6, ..., 2
r = np.array([0, 4, 15, 42, 60, 38, 25, 18, 10, 6, 3, 1, 0])   # mm/h

total = np.trapz(r, t)                              # ∫ r(t) dt  (mm)
print(f"total rainfall ≈ {total:.1f} mm over {t[-1]:.0f} hours")
print(f"average intensity ≈ {total/t[-1]:.1f} mm/h;  peak = {r.max()} mm/h")

# Volume of water falling on a 12 m x 10 m roof (1 mm on 1 m^2 = 1 litre)
roof = 12 * 10
print(f"rainwater on a {roof} m^2 roof ≈ {total*roof:,.0f} litres")
```

## Listing 7.4: Average value of a function - the average temperature of a day

```python
# Listing 7.4: Average value of a function - the average temperature of a day
# Illustrative model of the temperature in Tamale on a hot day (°C), t = hours after midnight:
#   T(t) = 30 + 6 sin(pi (t - 9) / 12)
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

t = sp.symbols("t", real=True)
T = 30 + 6*sp.sin(sp.pi*(t - 9)/12)

avg_day = sp.integrate(T, (t, 0, 24)) / 24
avg_work = sp.integrate(T, (t, 8, 18)) / (18 - 8)
print("average over the whole day       =", avg_day, "°C")
print("average from 8 am to 6 pm        ≈", round(float(avg_work), 2), "°C")

# Mean value theorem for integrals: T(c) equals the average at some time c
cs = sp.solve(sp.Eq(T, avg_work), t)
print("times when T equals the 8-6 average:", [round(float(c), 2) for c in cs if 8 <= c <= 18])

tt = np.linspace(0, 24, 300)
Tf = sp.lambdify(t, T)
plt.figure(figsize=(6, 3.2))
plt.plot(tt, Tf(tt), label="T(t)")
plt.axhline(float(avg_day), color="C1", ls="--", label="daily average")
plt.xlabel("hour of the day"); plt.ylabel("°C"); plt.grid(alpha=0.3); plt.legend()
plt.show()
```

## Listing 7.5: Volumes of revolution (disc and washer methods)

```python
# Listing 7.5: Volumes of revolution (disc and washer methods)
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

x, y, r, h = sp.symbols("x y r h", positive=True)

# (a) Sphere: rotate y = sqrt(r^2 - x^2) about the x-axis
print("sphere :", sp.simplify(sp.integrate(sp.pi*(r**2 - x**2), (x, -r, r))))
# (b) Cone: rotate y = (r/h) x, 0 <= x <= h, about the x-axis
print("cone   :", sp.integrate(sp.pi*(r*x/h)**2, (x, 0, h)))
# (c) Washer: region between y = 2x and y = x^2 rotated about the x-axis
print("washer :", sp.integrate(sp.pi*((2*x)**2 - (x**2)**2), (x, 0, 2)))

# (d) A clay water pot. Its inside radius (cm) at height y above the base is
#     R(y) = 12 + 8 sin(pi y / 40),  0 <= y <= 40   (illustrative shape)
R = 12 + 8*sp.sin(sp.pi*y/40)
V = sp.integrate(sp.pi*R**2, (y, 0, 40))
print("pot    :", sp.simplify(V), "≈", round(float(V)), "cm^3 ≈", round(float(V)/1000, 1), "litres")

Y = np.linspace(0, 40, 200)
Rf = sp.lambdify(y, R)
plt.figure(figsize=(4, 4))
plt.plot(Rf(Y), Y, "k"); plt.plot(-Rf(Y), Y, "k")
plt.axvline(0, ls="--", color="gray")
plt.gca().set_aspect("equal"); plt.title("profile of the pot (cm)"); plt.grid(alpha=0.3)
plt.show()
```

## Listing 7.6: Arc length L = ∫ sqrt(1 + (dy/dx)^2) dx

```python
# Listing 7.6: Arc length L = ∫ sqrt(1 + (dy/dx)^2) dx
import numpy as np
import sympy as sp
from scipy.integrate import quad

x = sp.symbols("x", positive=True)

# (a) An exact example: y = (2/3) x^(3/2) from x = 0 to x = 3
y = sp.Rational(2, 3)*x**sp.Rational(3, 2)
L = sp.integrate(sp.sqrt(1 + sp.diff(y, x)**2), (x, 0, 3))
print("y = (2/3)x^(3/2), 0 <= x <= 3:  L =", L)

# (b) A parabolic arch spanning 240 m and rising 50 m: y = 50 - k x^2
k = 50 / 120**2
dydx = lambda s: -2*k*s
L, _ = quad(lambda s: np.sqrt(1 + dydx(s)**2), -120, 120)
print(f"length of the arch = {L:.2f} m  (span 240 m, so {L - 240:.2f} m longer)")

# (c) A curve whose arc-length integral has no formula: y = sin x, 0 <= x <= pi
L, _ = quad(lambda s: np.sqrt(1 + np.cos(s)**2), 0, np.pi)
print(f"one arch of y = sin x has length {L:.6f}")
```

## Listing 7.7: Area of an irregular farm from a surveyor's offsets

```python
# Listing 7.7: Area of an irregular farm from a surveyor's offsets
# A straight baseline runs along the road edge of a farm. Every 20 m along it the
# surveyor measures the perpendicular width (offset) to the far boundary (illustrative).
import numpy as np
from scipy.integrate import simpson
import matplotlib.pyplot as plt

x = np.arange(0, 201, 20)                                   # 0, 20, ..., 200 m
w = np.array([0, 38, 52, 61, 70, 74, 69, 63, 55, 41, 0])    # offsets (m)

A_trap = np.trapz(w, x)
A_simp = simpson(w, x=x)
print(f"trapezium rule: {A_trap:8.0f} m^2 = {A_trap/10000:.3f} ha")
print(f"Simpson's rule: {A_simp:8.0f} m^2 = {A_simp/10000:.3f} ha")

plt.figure(figsize=(6, 2.8))
plt.fill_between(x, 0, w, alpha=0.3, color="green")
plt.plot(x, w, "go-"); plt.axhline(0, color="k", lw=3)
plt.text(80, -12, "road (baseline)")
plt.xlabel("distance along road (m)"); plt.ylabel("offset (m)")
plt.ylim(-20, 85); plt.grid(alpha=0.3); plt.show()
```

## Listing 7.8: Probability as area under a density curve

```python
# Listing 7.8: Probability as area under a density curve
import numpy as np
from scipy.integrate import quad
import matplotlib.pyplot as plt

# Examination marks modelled by a normal density with mean 55, standard deviation 12
mu, sigma = 55, 12
pdf = lambda x: np.exp(-((x - mu)/sigma)**2 / 2) / (sigma*np.sqrt(2*np.pi))

total, _ = quad(pdf, -np.inf, np.inf)
print(f"total area under the density = {total:.6f}")
for lo, hi in [(40, 70), (70, 100), (0, 40)]:
    p, _ = quad(pdf, lo, hi)
    print(f"P({lo} <= mark <= {hi}) ≈ {p:.3f}")

X = np.linspace(10, 100, 300)
plt.figure(figsize=(6, 3.2))
plt.plot(X, pdf(X), "k")
Xs = np.linspace(70, 100, 100)
plt.fill_between(Xs, pdf(Xs), alpha=0.4, label="P(mark ≥ 70)")
plt.xlabel("mark"); plt.grid(alpha=0.3); plt.legend()
plt.show()
```
