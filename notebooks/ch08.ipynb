## Listing 8.1: Plotting parametric curves x = f(t), y = g(t)

```python
# Listing 8.1: Plotting parametric curves x = f(t), y = g(t)
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

t = np.linspace(0, 2*np.pi, 400)
fig, ax = plt.subplots(1, 4, figsize=(10, 2.8))
ax[0].plot(np.cos(t), np.sin(t));           ax[0].set_title("circle")
ax[1].plot(2*np.cos(t), 3*np.sin(t), "C1");  ax[1].set_title("ellipse")
T = np.linspace(0, 4*np.pi, 400)
ax[2].plot(T - np.sin(T), 1 - np.cos(T), "C2"); ax[2].set_title("cycloid")
ax[3].plot(np.sin(3*t), np.sin(2*t), "C3");  ax[3].set_title("Lissajous")
for a in ax:
    a.set_aspect("equal"); a.grid(alpha=0.3)
plt.tight_layout(); plt.show()

# Eliminating the parameter with SymPy: x = 2 cos t, y = 3 sin t
s = sp.symbols("t")
x, y = 2*sp.cos(s), 3*sp.sin(s)
print("(x/2)^2 + (y/3)^2 =", sp.simplify((x/2)**2 + (y/3)**2))
```

## Listing 8.2: The path of a kicked football (no air resistance)

```python
# Listing 8.2: The path of a kicked football (no air resistance)
#   x = v cos(a) t,   y = v sin(a) t - g t^2 / 2
import numpy as np
import matplotlib.pyplot as plt

g, v = 9.8, 25.0                     # m/s^2, kick speed in m/s
plt.figure(figsize=(6.5, 3))
for deg in [20, 30, 45, 60]:
    a = np.radians(deg)
    T = 2*v*np.sin(a)/g              # time of flight (y = 0 again)
    t = np.linspace(0, T, 200)
    x, y = v*np.cos(a)*t, v*np.sin(a)*t - g*t**2/2
    print(f"angle {deg:>2}°: flight {T:4.2f} s, range {x[-1]:6.2f} m, "
          f"max height {(v*np.sin(a))**2/(2*g):5.2f} m")
    plt.plot(x, y, label=f"{deg}°")
plt.xlabel("x (m)"); plt.ylabel("y (m)"); plt.gca().set_aspect("equal")
plt.grid(alpha=0.3); plt.legend(fontsize=8); plt.show()

# A free kick 20 m from goal at 30°: how high is the ball when it reaches the goal line?
a = np.radians(30)
t_goal = 20 / (v*np.cos(a))
print(f"at the goal line: t = {t_goal:.3f} s, height = {v*np.sin(a)*t_goal - g*t_goal**2/2:.2f} m"
      " (crossbar 2.44 m)")
```

## Listing 8.3: Calculus of parametric curves - the cycloid x = r(t - sin t), y = r(1 - cos t)

```python
# Listing 8.3: Calculus of parametric curves - the cycloid x = r(t - sin t), y = r(1 - cos t)
import sympy as sp

t = sp.symbols("t", real=True)
r = sp.symbols("r", positive=True)
x, y = r*(t - sp.sin(t)), r*(1 - sp.cos(t))

dydx = sp.simplify(sp.diff(y, t) / sp.diff(x, t))
print("dy/dx =", dydx)
print("slope at t = pi/2:", dydx.subs(t, sp.pi/2))
d2 = sp.simplify(sp.diff(dydx, t) / sp.diff(x, t))
print("d2y/dx2 =", d2)

# Tangent line at t = pi/2 when r = 1
P = (x.subs({t: sp.pi/2, r: 1}), y.subs({t: sp.pi/2, r: 1}))
X = sp.symbols("X")
print("tangent at t = pi/2 (r = 1): y =", sp.expand(P[1] + dydx.subs(t, sp.pi/2)*(X - P[0])))

# Arc length of one arch: ∫_0^{2 pi} sqrt(x'^2 + y'^2) dt
speed = sp.sqrt(sp.diff(x, t)**2 + sp.diff(y, t)**2)
speed = 2*r*sp.sin(t/2)          # simplification of speed, valid for 0 <= t <= 2 pi
print("check speed formula:", sp.simplify((sp.diff(x, t)**2 + sp.diff(y, t)**2) - speed**2))
print("length of one arch =", sp.integrate(speed, (t, 0, 2*sp.pi)))
```

## Listing 8.4: Converting between polar and Cartesian coordinates

```python
# Listing 8.4: Converting between polar and Cartesian coordinates
import numpy as np

def to_cartesian(r, theta):
    return r*np.cos(theta), r*np.sin(theta)

def to_polar(x, y):
    return np.hypot(x, y), np.arctan2(y, x)       # arctan2 picks the correct quadrant

for r, deg in [(2, 30), (4, 135), (3, 270)]:
    x, y = np.round(to_cartesian(r, np.radians(deg)), 4) + 0.0   # + 0.0 removes "-0.0"
    print(f"(r, θ) = ({r}, {deg}°)  ->  (x, y) = ({x:.4f}, {y:.4f})")
for x, y in [(1, 1), (-1, 1), (-3, -4)]:
    r, th = to_polar(x, y)
    print(f"(x, y) = ({x}, {y})  ->  r = {r:.4f}, θ = {np.degrees(th):.2f}°  "
          f"(arctan(y/x) alone gives {np.degrees(np.arctan(y/x)):.2f}°)")

# An airport radar reports an aircraft at range 40 km on a BEARING of 120°
# (bearings are measured clockwise from north). Polar angle = 90° - bearing.
rng, bearing = 40, 120
x, y = to_cartesian(rng, np.radians(90 - bearing))
print(f"(x, y) = ({x:.2f}, {y:.2f}): the aircraft is {x:.2f} km east and "
      f"{-y:.2f} km south of the radar")
```

## Listing 8.5: A gallery of polar curves r = f(theta)

```python
# Listing 8.5: A gallery of polar curves r = f(theta)
import numpy as np
import matplotlib.pyplot as plt

th = np.linspace(0, 2*np.pi, 1000)
curves = {
    "r = 2":                  2 + 0*th,
    "r = 2 cos θ":            2*np.cos(th),
    "r = 1 + cos θ (cardioid)": 1 + np.cos(th),
    "r = cos 3θ (3 petals)":  np.cos(3*th),
    "r = cos 2θ (4 petals)":  np.cos(2*th),
    "r = θ/π (spiral)":       th/np.pi,
}
fig, ax = plt.subplots(2, 3, figsize=(9, 6), subplot_kw={"projection": "polar"})
for a, (name, r) in zip(ax.flat, curves.items()):
    a.plot(th, r)                    # matplotlib's polar axes accept negative r
    a.set_title(name, fontsize=9)
    a.set_yticklabels([])
plt.tight_layout(); plt.show()
```

## Listing 8.6: Slopes and areas in polar coordinates

```python
# Listing 8.6: Slopes and areas in polar coordinates
import sympy as sp

th = sp.symbols("theta", real=True)
a = sp.symbols("a", positive=True)

# Slope of r = f(theta):  dy/dx = (r' sin + r cos) / (r' cos - r sin)
r = 1 + sp.cos(th)                                  # cardioid
rp = sp.diff(r, th)
slope = sp.simplify((rp*sp.sin(th) + r*sp.cos(th)) / (rp*sp.cos(th) - r*sp.sin(th)))
print("cardioid slope at theta = pi/2:", slope.subs(th, sp.pi/2))

# Area A = (1/2) ∫ r^2 dtheta
print("area of cardioid r = a(1 + cos θ):", sp.integrate((a*(1 + sp.cos(th)))**2/2, (th, 0, 2*sp.pi)))
print("one petal of r = cos 2θ          :", sp.integrate(sp.cos(2*th)**2/2, (th, -sp.pi/4, sp.pi/4)))
print("whole circle r = 2a cos θ        :", sp.integrate((2*a*sp.cos(th))**2/2, (th, -sp.pi/2, sp.pi/2)))

# Inside r = 3 sin θ but outside r = 1 + sin θ: the curves meet where sin θ = 1/2
area = sp.integrate((9*sp.sin(th)**2 - (1 + sp.sin(th))**2)/2, (th, sp.pi/6, 5*sp.pi/6))
print("inside r = 3 sin θ, outside r = 1 + sin θ:", sp.simplify(area))

# Arc length in polar form: L = ∫ sqrt(r^2 + r'^2) dθ for the cardioid r = 1 + cos θ
L = sp.integrate(2*sp.cos(th/2), (th, 0, sp.pi)) * 2      # sqrt(r^2 + r'^2) = 2|cos(θ/2)|
print("length of the cardioid r = 1 + cos θ:", L)
```
