## Listing 9.1: A parabola is the set of points equidistant from a focus and a directrix

```python
# Listing 9.1: A parabola is the set of points equidistant from a focus and a directrix
import numpy as np
import matplotlib.pyplot as plt

c = 2                                   # x^2 = 4cy = 8y: focus (0, 2), directrix y = -2
x = np.linspace(-6, 6, 13)
y = x**2 / (4*c)
d_focus = np.sqrt(x**2 + (y - c)**2)
d_directrix = y + c
print(" x      y   dist to focus   dist to directrix")
for row in zip(x[::3], y[::3], d_focus[::3], d_directrix[::3]):
    print("{:>3.0f} {:>6.2f} {:>15.4f} {:>19.4f}".format(*row))

X = np.linspace(-6, 6, 200)
plt.figure(figsize=(5, 3.8))
plt.plot(X, X**2/(4*c), "k", lw=2, label="$x^2 = 8y$")
plt.axhline(-c, color="C3", ls="--", label="directrix $y=-2$")
plt.plot(0, c, "C3*", ms=12, label="focus (0, 2)")
P = (4, 2)                              # a sample point on the parabola
plt.plot([P[0], 0], [P[1], c], "C0"); plt.plot([P[0], P[0]], [P[1], -c], "C0")
plt.plot(*P, "C0o")
plt.gca().set_aspect("equal"); plt.ylim(-3, 6); plt.grid(alpha=0.3); plt.legend(fontsize=8)
plt.show()
```

## Listing 9.2: Where should the receiver of a satellite dish go?

```python
# Listing 9.2: Where should the receiver of a satellite dish go?
import numpy as np
import matplotlib.pyplot as plt

D, depth = 90.0, 12.0                  # dish diameter and depth (cm)
# Dish cross-section: x^2 = 4cy passes through (D/2, depth)
c = (D/2)**2 / (4*depth)
print(f"focal length c = D^2/(16 depth) = {c:.2f} cm above the centre of the dish")

# Reflect vertical incoming rays off y = x^2/(4c): they all pass through (0, c)
X = np.linspace(-D/2, D/2, 200)
plt.figure(figsize=(5, 4))
plt.plot(X, X**2/(4*c), "k", lw=3)
for x0 in np.linspace(-40, 40, 9):
    y0 = x0**2/(4*c)
    plt.plot([x0, x0], [60, y0], "C1", lw=0.8)            # incoming ray
    plt.plot([x0, 0], [y0, c], "C0", lw=0.8)              # reflected ray
plt.plot(0, c, "r*", ms=14)
plt.gca().set_aspect("equal"); plt.ylim(-2, 62); plt.grid(alpha=0.3)
plt.title("all reflected rays meet at the focus")
plt.show()
```

## Listing 9.4: Analysing the ellipse 3x^2 + 4y^2 + 24x - 16y + 52 = 0

```python
# Listing 9.4: Analysing the ellipse 3x^2 + 4y^2 + 24x - 16y + 52 = 0
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

x, y = sp.symbols("x y")
E = 3*x**2 + 4*y**2 + 24*x - 16*y + 52
# Completing the square: 3(x+4)^2 + 4(y-2)^2 = 12
print("check:", sp.expand(3*(x + 4)**2 + 4*(y - 2)**2 - 12 - E) == 0)
h, k, a, b = -4, 2, 2, np.sqrt(3)
c = np.sqrt(a**2 - b**2)
print(f"centre ({h}, {k}), a = {a}, b = {b:.4f}, c = {c:.0f}, e = c/a = {c/a}")
F1, F2 = np.array([h - c, k]), np.array([h + c, k])
print("foci:", F1, F2)

t = np.linspace(0, 2*np.pi, 7)[:-1]
for P in zip(h + a*np.cos(t), k + b*np.sin(t)):          # sample points on the ellipse
    P = np.array(P)
    print(f"P = ({P[0]:6.3f}, {P[1]:6.3f}):  PF1 + PF2 = "
          f"{np.linalg.norm(P - F1) + np.linalg.norm(P - F2):.6f}")

T = np.linspace(0, 2*np.pi, 300)
plt.figure(figsize=(5, 3.4))
plt.plot(h + a*np.cos(T), k + b*np.sin(T), "k")
plt.plot(*F1, "r*", ms=10); plt.plot(*F2, "r*", ms=10)
plt.gca().set_aspect("equal"); plt.grid(alpha=0.3); plt.show()
```

## Listing 9.5: Elliptical orbits - the Earth, and Ghana's first satellite

```python
# Listing 9.5: Elliptical orbits - the Earth, and Ghana's first satellite
import numpy as np
from scipy.integrate import quad

# Earth's orbit: closest to the Sun 147.1 million km, furthest 152.1 million km
peri, aph = 147.1, 152.1
a = (peri + aph) / 2                  # semi-major axis
c = a - peri                          # centre-to-focus distance (Sun at a focus)
e = c / a
b = np.sqrt(a**2 - c**2)
print(f"a = {a:.2f}, b = {b:.3f} (million km), e = {e:.4f}")

# Length of the orbit: arc length of x = a cos t, y = b sin t (an integral, Ch. 6)
L, _ = quad(lambda t: np.sqrt((a*np.sin(t))**2 + (b*np.cos(t))**2), 0, 2*np.pi)
print(f"orbit length = {L:.2f} million km;  2*pi*a = {2*np.pi*a:.2f}")
print(f"average orbital speed = {L*1e6/(365.256*24*3600):.1f} km/s")

# GhanaSat-1 orbited about 400 km above the Earth (radius 6371 km).
# Kepler's third law: period T = 2 pi sqrt(a^3 / GM), GM = 398600 km^3/s^2
GM, a_sat = 398600, 6371 + 400
T = 2*np.pi*np.sqrt(a_sat**3 / GM)
print(f"period at 400 km altitude = {T/60:.1f} minutes, i.e. {24*3600/T:.1f} orbits per day")
```

## Listing 9.6: The hyperbola 16x^2 - 9y^2 - 128x - 18y + 103 = 0 and its asymptotes

```python
# Listing 9.6: The hyperbola 16x^2 - 9y^2 - 128x - 18y + 103 = 0 and its asymptotes
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

x, y = sp.symbols("x y")
H = 16*x**2 - 9*y**2 - 128*x - 18*y + 103
standard = 16*(x - 4)**2 - 9*(y + 1)**2 - 144
print("16(x-4)^2 - 9(y+1)^2 = 144 is the same curve:", sp.expand(standard - H) == 0)
h, k, a, b = 4, -1, 3, 4
c = np.sqrt(a**2 + b**2)
print(f"centre ({h}, {k}), a = {a}, b = {b}, c = {c:.0f}, e = {c/a:.4f}")
print(f"foci ({h - c:.0f}, {k}) and ({h + c:.0f}, {k}); asymptotes y + 1 = ±(4/3)(x - 4)")

t = np.linspace(-2, 2, 200)                       # x = h ± a cosh t, y = k + b sinh t
plt.figure(figsize=(5, 4))
plt.plot(h + a*np.cosh(t), k + b*np.sinh(t), "k")
plt.plot(h - a*np.cosh(t), k + b*np.sinh(t), "k")
X = np.linspace(-6, 14, 2)
plt.plot(X, k + b/a*(X - h), "C3--", lw=1); plt.plot(X, k - b/a*(X - h), "C3--", lw=1)
plt.plot([h - c, h + c], [k, k], "b*", ms=10)
plt.xlim(-6, 14); plt.ylim(-12, 10); plt.gca().set_aspect("equal"); plt.grid(alpha=0.3)
plt.show()
```

## Listing 9.7: Locating a lightning strike from arrival-time differences (hyperbolas)

```python
# Listing 9.7: Locating a lightning strike from arrival-time differences (hyperbolas)
import numpy as np
from scipy.optimize import fsolve

v = 3.0e5                                           # radio waves travel at 300 000 km/s
S = np.array([[0.0, 0.0], [40.0, 0.0], [10.0, 35.0]])   # three weather stations (km)

true = np.array([25.0, 18.0])                       # used only to simulate the data
arrival = np.linalg.norm(S - true, axis=1) / v      # arrival times (s)
dt12, dt13 = arrival[0] - arrival[1], arrival[0] - arrival[2]
print(f"time differences: {dt12*1e6:.3f} and {dt13*1e6:.3f} microseconds")

# Each time difference puts the strike on one branch of a hyperbola with foci at
# two stations:  |PS1| - |PS2| = v * dt12  and  |PS1| - |PS3| = v * dt13
def equations(P):
    d = np.linalg.norm(S - P, axis=1)
    return [d[0] - d[1] - v*dt12, d[0] - d[2] - v*dt13]

estimate = fsolve(equations, x0=[20, 20])
print("estimated strike position:", np.round(estimate, 4), "km")
```

## Listing 9.8: Classifying Ax^2 + Bxy + Cy^2 + Dx + Ey + F = 0 and plotting it

```python
# Listing 9.8: Classifying Ax^2 + Bxy + Cy^2 + Dx + Ey + F = 0 and plotting it
import numpy as np
import matplotlib.pyplot as plt

def classify(A, B, C):
    disc = B**2 - 4*A*C
    if disc < 0:
        return "circle" if (A == C and B == 0) else "ellipse"
    return "parabola" if disc == 0 else "hyperbola"

curves = {                                     # (A, B, C, D, E, F)
    "x²+y²+3x−6y−1=0":         (1, 0, 1, 3, -6, -1),
    "3x²+4y²+24x−16y+52=0":    (3, 0, 4, 24, -16, 52),
    "x²+4y+8x+4=0":            (1, 0, 0, 8, 4, 4),
    "16x²−9y²−128x−18y+103=0": (16, 0, -9, -128, -18, 103),
    "x²+2xy+y²+6x−34y−71=0":   (1, 2, 1, 6, -34, -71),
}
X, Y = np.meshgrid(np.linspace(-12, 16, 600), np.linspace(-12, 16, 600))
plt.figure(figsize=(5.5, 5.5))
for i, (name, (A, B, C, D, E, F)) in enumerate(curves.items()):
    kind = classify(A, B, C)
    print(f"{name:<26} B²−4AC = {B**2 - 4*A*C:>5}  -> {kind}")
    Z = A*X**2 + B*X*Y + C*Y**2 + D*X + E*Y + F
    plt.contour(X, Y, Z, levels=[0], colors=f"C{i}")
    plt.plot([], [], color=f"C{i}", label=f"{kind}")
plt.gca().set_aspect("equal"); plt.grid(alpha=0.3); plt.legend(fontsize=8, loc="lower left")
plt.show()
```

## Listing 9.9: One equation for all conics: r = e d / (1 + e cos(theta))

```python
# Listing 9.9: One equation for all conics: r = e d / (1 + e cos(theta))
import numpy as np
import matplotlib.pyplot as plt

d = 1.0                                      # distance from the focus to the directrix
theta = np.linspace(0, 2*np.pi, 2000)
plt.figure(figsize=(5.5, 4.5))
for e, style in [(0.5, "C0"), (0.8, "C2"), (1.0, "C3"), (1.5, "C1")]:
    r = e*d / (1 + e*np.cos(theta))
    r[(1 + e*np.cos(theta)) <= 0.02] = np.nan         # skip points at infinity
    plt.plot(r*np.cos(theta), r*np.sin(theta), style, label=f"e = {e}")
plt.axvline(d, color="k", ls="--", lw=1, label="directrix")
plt.plot(0, 0, "k*", ms=10, label="focus")
plt.xlim(-4, 1.5); plt.ylim(-3, 3); plt.gca().set_aspect("equal")
plt.grid(alpha=0.3); plt.legend(fontsize=8)
plt.show()
```
